```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv("Expanded_data_with_more_features.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.drop(columns='Unnamed: 0',inplace=True)
```

```python
df.head()
```

```python
plt.figure(figsize=(5,5))
ax=sns.countplot(data=df,x="Gender")
ax.bar_label(ax.containers[0])
plt.show()
```

```python
gp=df.groupby('ParentEduc').agg({"MathScore":'mean',"ReadingScore":"mean","WritingScore":"mean"})
gp
```

```python
#from the below chart the parents has the maximum impact on educations of their wards depends on the education  level
```

```python
plt.figure(figsize=(4,4))
sns.heatmap(gp,annot=True)
plt.show()
```

```python
gb=df.groupby("ParentMaritalStatus").agg({"MathScore":'mean',"ReadingScore":'mean',"WritingScore":'mean'})
gb
```

```python
plt.figure(figsize=(4,4),)
sns.heatmap(gb,annot=True)
plt.show()
```
